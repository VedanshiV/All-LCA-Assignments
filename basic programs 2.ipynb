```python
#calculte percentage and display highest and lowest
 #accept student details
name = input("Enter student name: ")
roll_no = input("Enter roll number: ")
 #accept marks for three subjects
subject1 = input("Enter subject 1 name: ")
marks1 = float(input(f"Enter marks in {subject1}: "))

subject2 = input("Enter subject 2 name: ")
marks2 = float(input(f"Enter marks in {subject2}: "))

subject3 = input("Enter subject 3 name: ")
marks3 = float(input(f"Enter marks in {subject3}: "))

 #calculate percentage
total = marks1 + marks2 + marks3
percentage = (total/300) * 100

 #display student details and percentage
print("\n--- Student Result ---")
print("Student Name:", name)
print("Roll Number:", roll_no)
print("Percentage:", percentage)

 #to find the highest and lowest marks
if marks1 >= marks2 and marks1 >= marks3:
    print("Subject with highest marks:", subject1)
elif marks2 >= marks1 and marks2 >= marks3:
    print("Subject with highest marks:", subject2)
else:
    print("Subject with highest marks:", subject3)

if marks1 <= marks2 and marks1 <= marks3:
    print("Subject with lowest marks:", subject1)
elif marks2 <= marks1 and marks2 <= marks3:
    print("Subject with lowest marks:", subject2)
else:
    print("Subject with lowest marks:", subject3)
    
```

```python
# digit reversal
num = int(input("enter a number: "))
reverse = 0
while num >0:
 digit = num % 10
 reverse = reverse * 10 + digit
 num = num // 10
print("reverse =", reverse)
```

```python
#leap year
year = int(input("Enter a year: "))
if year % 400 == 0:
    print("the year is a leap year")
elif year % 100 == 0:
    print("the year is not a leap year")
elif year % 4 == 0:
    print("the year is a leap year")
else:
    print("the year is not a leap year")
```
