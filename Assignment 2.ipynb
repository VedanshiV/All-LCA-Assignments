```python
# Assignment 2:
#  To find largest of three numbers

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a == b == c:
    print("All three numbers are equal:", a)
elif a == b and a > c:
    print("Largest numbers:", a)
elif a == c and a > b:
    print("Largest numbers:", a)
elif b == c and b > a:
    print("Largest numbers:", b)
elif a > b and a > c:
    print("Largest:", a)
elif b > a and b > c:
    print("Largest:", b)
else:
    print("Largest:", c)
```
