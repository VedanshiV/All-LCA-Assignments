```python
# code with output(colab)
import re

def check_string(s):
    pattern = "^[A-Za-z0-9]+$"
    # ^ start, $ end, + one or more
    return bool(re.match(pattern,s))


# example usage
s1 = "Hello123"
s2 = "Hello_123"

print(check_string(s1)) # true
print(check_string(s2)) # false
```

```python
# another way to complete the assignment
def check_string(s):
    return s.isalnum()
    
# example usage
s1 = "Hello123"
s2 = "Hello_123"

print(check_string(s1)) # true
print(check_string(s2)) # false
```
